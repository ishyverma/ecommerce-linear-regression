## Feature and Target Identification

The target variable is `Yearly Amount Spent`.

The numerical variables that may be used as predictors are:

- Avg. Session Length
- Time on App
- Time on Website
- Length of Membership

The columns `Email`, `Address`, and `Avatar` will not be used as predictors
in the initial Linear Regression model.

```python
df[numerical_columns].hist(
    figsize=(12, 8),
    bins=20
)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x='Yearly Amount Spent',
    kde=True
)

plt.title("Distribution of Yearly Amount Spent")
plt.xlabel("Yearly Amount Spent")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 4))

sns.boxplot(
    x=df['Yearly Amount Spent']
)

plt.title("Boxplot of Yearly Amount Spent")
plt.xlabel("Yearly Amount Spent")

plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

features = ["Avg. Session Length", "Time on App", "Time on Website", "Length of Membership"]

for ax, feature in zip(axes.flatten(), features):
    sns.histplot(data=df, x=feature, kde=True, ax=ax)
    ax.set_title(f"Distribution of {feature}")

plt.tight_layout()
plt.show()
```

```python
df[numerical_columns].skew()
```

```python
correlation_matrix = df[numerical_columns].corr()

correlation_matrix
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Matrix")
plt.show()
```

```python
target_correlation = (
    df[numerical_columns]
    .corr()["Yearly Amount Spent"]
    .sort_values(ascending=False)
)

target_correlation
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Avg. Session Length",
    y="Yearly Amount Spent"
)

plt.title("Avg. Session Length vs Yearly Amount Spent")
plt.xlabel("Avg. Session Length")
plt.ylabel("Yearly Amount Spent")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Time on App",
    y="Yearly Amount Spent"
)

plt.title("Time on App vs Yearly Amount Spent")
plt.xlabel("Time on App")
plt.ylabel("Yearly Amount Spent")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Time on Website",
    y="Yearly Amount Spent"
)

plt.title("Time on Website vs Yearly Amount Spent")
plt.xlabel("Time on Website")
plt.ylabel("Yearly Amount Spent")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Length of Membership",
    y="Yearly Amount Spent"
)

plt.title("Length of Membership vs Yearly Amount Spent")
plt.xlabel("Length of Membership")
plt.ylabel("Yearly Amount Spent")

plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

features = ["Avg. Session Length", "Time on App", "Time on Website", "Length of Membership"]

for ax, feature in zip(axes.flatten(), features):
    sns.regplot(
        data=df,
        x=feature,
        y="Yearly Amount Spent",
        scatter_kws={ "alpha": 0.5 },
        ax=ax
    )

    ax.set_title(f"{feature} vs Yearly Amount Spent")

plt.tight_layout()
plt.show()
```

```python
sns.pairplot(
    df[["Avg. Session Length", "Time on App", "Time on Website", "Length of Membership", "Yearly Amount Spent"]]
)

plt.show()
```

# EDA Summary

## Dataset
- The dataset contains 500 observations and 8 columns.
- There are 5 numerical variables and 3 object-type variables.
- `Yearly Amount Spent` is the target variable for the regression problem.

## Data Quality
- No missing values were found.
- No duplicate rows were found.
- The numerical variables contain valid numerical observations.

## Variables
- `Email` and `Address` are identifiers/information fields and will not be used as model predictors.
- `Avatar` will also be excluded from the initial regression model.
- The four numerical behavioral variables will be considered as predictors.

## Relationship Analysis
- Scatter plots and correlation analysis were used to examine the relationship
  between each numerical predictor and `Yearly Amount Spent`.
- The relationships appear to be approximately linear to varying degrees.
- Correlation values provide information about linear association but do not
  establish causation.
